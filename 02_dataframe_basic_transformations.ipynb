```python
from pyspark.sql.functions import *

df = (spark.read
      .format("csv")
      .option("header", True)
      .option("inferSchema", True)
      .load("/Volumes/workspace/default/pyspark_files/BigMart Sales.csv"))

display(df)
```

```python
df.printSchema()
```

### SELECT->
#####Pulls specific columns. Can use string names or the `col()` object for advanced operations.

```python
from pyspark.sql.functions import col

#df.select('Item_Identifier', 'Item_Weight', 'Item_MRP').display() #just for selecting
df.select(col('Item_Identifier'), col('Item_Weight'), col('Item_MRP')).display() #to use functions too
```

###ALIAS
#####Renames a column within a transformation

```python
df.select(col('Item_Identifier').alias('Item_ID')).display()
```

###FILTER
#####Slices data based on conditions. Multiple conditions use boolean operators: `&amp;` (And), `|` (Or), `~` (Not).

```python
from pyspark.sql.functions import col

#scenario 1
df.filter(col('Item_Fat_Content') == 'Regular').display()
```

```python
#scenario 2 (multiple conditions)
df.filter((col('Item_Type') == 'Soft Drinks') & (col('Item_MRP') < 100)).display()
```

```python
#scenario-3

df.filter(col('Outlet_size').isNull() & col('Outlet_Location_Type').isin('Tier 1', 'Tier 2')).display()
```

### Renaming a Column using (withColumnRenamed()) 
#####Specifically for changing a column name at the DataFrame level.

```python
df.withColumnRenamed('Item_Weight', 'Item_Wt').display()
```

### Adding a new column or Modifying exsisting using (withColumn)
#####Used to create a new column or modify an existing one. Constant values require the `lit()` function.

```python
from pyspark.sql.functions import lit

df = df.withColumn('flag', lit("new")) #adding new col with constant value
display(df)
```

```python
df.withColumn('multiply', col('Item_Weight')*col("Item_MRP")).display()
```

```python
from pyspark.sql.functions import regexp_replace

df.withColumn('Item_Fat_content', regexp_replace(col('Item_Fat_Content'), "Regular", 'Reg'))\
    .withColumn('Item_Fat_content', regexp_replace(col('Item_Fat_content'), "Low Fat", 'LF')).display()
```

### Type Casting (changing datatypes)
#####Changes the data type of a column

```python
df.withColumn('Item_Weight', col("Item_Weight").cast(StringType())).display()
```

###Sort/orderBy
#####Arranges data. Use `.asc()` for ascending and `.desc()` for descending.

```python
df.sort(col('Item_Weight').desc()).display() #asc -> ascending
```

```python
df.sort(['Item_Weight', 'Item_Visibility'], ascending=[0, 1]).display()
```

### Limit (same as sql)

```python
df.limit(10).display()
```
