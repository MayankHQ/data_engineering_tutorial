## Window Functions

Window functions perform calculations across a "window" of rows related to the current row.

### Ranking Functions

* **row\_number** **:** Assigns a unique, sequential number.
* **rank** **:** Assigns ranks but leaves gaps if there are ties (e.g., 1, 1, 3).
* **dense\_rank** **:** Assigns ranks without gaps (e.g., 1, 1, 2).

```python
from pyspark.sql.window import Window

df1 = df.withColumn('unique_row_num', row_number().over(Window.orderBy('Item_Identifier')))
df1.display()
```

```python
df1.withColumn('rank', rank().over(Window.orderBy(col('Item_Identifier'))))\
    .withColumn('dense_rank', dense_rank().over(Window.orderBy(col('Item_Identifier')))).display()
```
